# Previsão de Séries Temporais *Zero-Shot* com Amazon Chronos (T5 & Bolt) vs. Modelos Clássicos: Consumo de Energia Elétrica no Brasil (EPE)

---

## 1. O que são *Foundation Models* de Séries Temporais (Amazon Chronos)?

Modelos tradicionais de séries temporais — como **ARIMA/SARIMA** ou **Holt-Winters (ETS)** — precisam ser ajustados do zero para cada série histórica individual. Já o **Amazon Chronos** inaugura uma nova geração: **Modelos Fundacionais Pré-Treinados (*Zero-Shot*) para Séries Temporais**.

### Como o Chronos funciona de forma intuitiva?
Pense no Chronos como um **"autocompletar de números"** baseado na mesma arquitetura *Transformer* dos grandes modelos de linguagem (LLMs):
1. **Escalonamento e Tokenização**: Ele normaliza os valores históricos da série (dividindo pela média absoluta) e os converte em *tokens* discretos — tratando os números como se fossem "palavras de uma frase".
2. **Pré-Treinamento Massivo**: O modelo já foi treinado pela Amazon Science em **milhões de séries temporais reais e sintéticas**, aprendendo padrões universais de tendência, sazonalidade e choques.
3. **Inferência *Zero-Shot* Probabilística**: Quando entregamos uma série nova que o Chronos nunca viu (como o consumo de energia do Brasil), ele projeta diretamente o futuro em forma de **distribuição probabilística**, gerando a mediana (**P50**) e faixas de incerteza (**P10–P90**).

### Diferença Prática: Família `Chronos-T5` vs. `Chronos-Bolt`
- **`Chronos-T5`** (1ª geração): Arquitetura *encoder-decoder* T5 que gera previsões amostrando múltiplas trajetórias autoregressivas passo a passo — excelente qualidade, porém mais lento em CPU.
- **`Chronos-Bolt`** (2ª geração): Baseado em *patching* direto de múltiplos quantis (*direct multi-step quantile forecasting*) — **até dezenas de vezes mais rápido em CPU**, consumindo menos memória e frequentemente com acurácia igual ou superior!

Utilizamos a base oficial mensal de **Consumo de Energia Elétrica no Brasil (EPE — Empresa de Pesquisa Energética)** em `dados/consumo_energia_pronto.xlsx`, cobrindo **259 meses (Janeiro/2004 a Julho/2025)** com 63 séries agregadas por região (`N`, `NE`, `CO`, `SE`, `S`) e classe (`residencial`, `comercial`, `industrial`, `rural`, `outros`).

## 3. Passo 1: Carregamento e Exploração do Consumo de Energia do Brasil (EPE)

### O que vamos fazer agora e por quê?
Carregamos `dados/consumo_energia_pronto.xlsx`, convertemos a coluna `Data` para índice mensal e expressamos a variável principal `ec_total_BR` (Consumo Total Brasil em MWh) em **GWh (`ec_total_BR / 1.000`)** para facilitar a leitura das métricas de erro (MAE e RMSE). Em seguida, visualizamos a evolução de 21 anos e a assinatura sazonal mês a mês.

```python
import inspect
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import seaborn as sns

from chronos import BaseChronosPipeline
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, mean_absolute_percentage_error, r2_score

warnings.filterwarnings("ignore")
pd.options.display.float_format = "{:.2f}".format
sns.set_theme(style="whitegrid")
plt.rcParams["figure.dpi"] = 110

# 1. Leitura da base preparada da EPE
df = pd.read_excel("dados/consumo_energia_pronto.xlsx")
df["Data"] = pd.to_datetime(df["Data"].astype(str), format="%Y%m%d")
df = df.sort_values("Data").reset_index(drop=True)

# Conversão de MWh para GWh (1 GWh = 1.000 MWh) para leitura executiva clara
df["consumo_BR_GWh"] = df["ec_total_BR"] / 1_000.0
df["Mes"] = df["Data"].dt.month
df["Ano"] = df["Data"].dt.year

print("=== SÉRIE OFICIAL DE CONSUMO DE ENERGIA ELÉTRICA — BRASIL (EPE) ===")
print(f"• Período coberto  : {df['Data'].min().strftime('%Y-%m')} até {df['Data'].max().strftime('%Y-%m')} ({len(df)} meses)")
print(f"• Séries presentes : {df.shape[1] - 3} variáveis (regiões e classes de consumo)")
print(f"• Consumo em Jan/2004 : {df['consumo_BR_GWh'].iloc[0]:,.1f} GWh")
print(f"• Consumo em Jul/2025 : {df['consumo_BR_GWh'].iloc[-1]:,.1f} GWh")
print(f"• Crescimento acumulado (2004–2025): {(df['consumo_BR_GWh'].iloc[-1]/df['consumo_BR_GWh'].iloc[0] - 1)*100:+.1f}%")

# 2. Visualização exploratória: Série histórica + Perfil Sazonal Mensal
fig, axes = plt.subplots(1, 2, figsize=(15, 4.8), gridspec_kw={"width_ratios": [2.2, 1.0]})

axes[0].plot(df["Data"], df["consumo_BR_GWh"], color="#1f77b4", lw=2.0)
axes[0].axvspan(pd.Timestamp("2020-03-01"), pd.Timestamp("2020-12-01"), color="#d62728", alpha=0.14, label="Choque Pandemia (2020)")
axes[0].axvspan(df["Data"].iloc[-12], df["Data"].iloc[-1], color="#2ca02c", alpha=0.14, label="Janela de Backtest (Últimos 12 meses)")
axes[0].set_title("Consumo Mensal Total de Energia Elétrica no Brasil (2004–2025)", fontsize=12.5, fontweight="bold")
axes[0].set_xlabel("Data")
axes[0].set_ylabel("Consumo Total (GWh)")
axes[0].legend(loc="upper left", frameon=True)

# Perfil sazonal médio dos últimos 5 anos
sazonal = df[df["Ano"] >= 2020].groupby("Mes")["consumo_BR_GWh"].mean()
meses_lbl = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
axes[1].bar(meses_lbl, sazonal.values, color="#2b5c8f", alpha=0.88, edgecolor="black")
axes[1].set_title("Perfil Sazonal Médio (2020–2025)", fontsize=12.5, fontweight="bold")
axes[1].set_xlabel("Mês do Ano")
axes[1].set_ylabel("Média Mensal (GWh)")
axes[1].set_ylim(sazonal.min() * 0.90, sazonal.max() * 1.05)
axes[1].tick_params(axis="x", rotation=35)

plt.tight_layout()
plt.show()
```

### O que a série nos revela antes da modelagem?
- **Forte tendência secular de crescimento**: O consumo nacional saltou de **~27.000 GWh/mês** (2004) para **~45.000–49.000 GWh/mês** (2024–2025), um avanço superior a **+67%**.
- **Sazonalidade térmica bem marcada**: Picos claros no **verão (Jan–Mar)** devido ao uso intensivo de refrigeração/ar-condicionado, e vales no **inverno (Jun–Jul)**.
- **Regimes e choques exógenos**: Observa-se uma queda abrupta na pandemia (2020) seguida por aceleração pós-2023 (ondas de calor no Brasil).

## 4. Passo 2: Split Temporal e Baselines Clássicos (Seasonal Naive & Holt-Winters)

### O que vamos fazer agora e por quê?
Todo benchmark sério de *Foundation Models* precisa responder a uma pergunta objetiva: **"O modelo de Deep Learning pré-treinado realmente supera métodos estatísticos clássicos rápidos e baratos?"**

Dividimos os 259 meses em:
- **Treino / Contexto (`247 meses`)**: Jan/2004 a Jul/2024.
- **Teste / Backtest (`12 meses`)**: Ago/2024 a Jul/2025.

E avaliamos dois baselines fundamentais:
1. **Seasonal Naive ($t - 12$)**: Repete exatamente o valor do mesmo mês no ano anterior.
2. **Holt-Winters Multiplicativo (`ExponentialSmoothing`)**: Modelo clássico de suavização exponencial com tendência aditiva e sazonalidade multiplicativa de 12 meses (usando erros de previsão para gerar a banda de confiança P10–P90).

```python
TEST_HORIZON = 12
train_df = df.iloc[:-TEST_HORIZON].copy()
test_df = df.iloc[-TEST_HORIZON:].copy()

y_train = train_df["consumo_BR_GWh"].values.astype("float64")
y_test = test_df["consumo_BR_GWh"].values.astype("float64")
datas_test = test_df["Data"].values

def avaliar_previsao(nome, familia, y_real, p50, p10=None, p90=None, tempo_s=0.0):
    '''Calcula métricas determinísticas e cobertura probabilística P10-P90.'''
    mae = mean_absolute_error(y_real, p50)
    rmse = root_mean_squared_error(y_real, p50)
    mape = mean_absolute_percentage_error(y_real, p50) * 100.0
    smape = np.mean(2.0 * np.abs(p50 - y_real) / (np.abs(y_real) + np.abs(p50))) * 100.0
    r2 = r2_score(y_real, p50)
    if p10 is not None and p90 is not None:
        cobertura = np.mean((y_real >= p10) & (y_real <= p90)) * 100.0
        largura_media = float(np.mean(p90 - p10))
    else:
        cobertura = np.nan
        largura_media = np.nan
    return {
        "Modelo": nome,
        "Familia": familia,
        "MAPE (%)": round(mape, 2),
        "sMAPE (%)": round(smape, 2),
        "MAE (GWh)": round(mae, 1),
        "RMSE (GWh)": round(rmse, 1),
        "R2": round(r2, 4),
        "Cobertura P10-P90 (%)": round(cobertura, 1) if not np.isnan(cobertura) else np.nan,
        "Tempo CPU (s)": round(tempo_s, 2),
    }

resultados_bench = []
previsoes_teste = {}

# 1. Baseline 1: Seasonal Naive (mesmo mês do ano anterior)
t0 = time.perf_counter()
pred_snaive = y_train[-12:]
dt_snaive = time.perf_counter() - t0
previsoes_teste["Seasonal Naive (t-12)"] = {"p50": pred_snaive, "p10": None, "p90": None}
resultados_bench.append(avaliar_previsao("Seasonal Naive (t-12)", "Baseline Clássico", y_test, pred_snaive, tempo_s=dt_snaive))

# 2. Baseline 2: Holt-Winters Multiplicativo (ETS)
t0 = time.perf_counter()
hw_model = ExponentialSmoothing(
    y_train, trend="add", seasonal="mul", seasonal_periods=12, damped_trend=False
).fit(optimized=True)
pred_hw = hw_model.forecast(TEST_HORIZON)
resid_std = np.std(hw_model.resid)
# Aproximação de intervalo P10-P90 (z = 1.2816) crescendo com sqrt(h)
h_steps = np.sqrt(np.arange(1, TEST_HORIZON + 1))
hw_p10 = pred_hw - 1.2816 * resid_std * h_steps
hw_p90 = pred_hw + 1.2816 * resid_std * h_steps
dt_hw = time.perf_counter() - t0

previsoes_teste["Holt-Winters (ETS)"] = {"p50": pred_hw, "p10": hw_p10, "p90": hw_p90}
resultados_bench.append(avaliar_previsao("Holt-Winters (ETS)", "Baseline Clássico", y_test, pred_hw, hw_p10, hw_p90, tempo_s=dt_hw))

print(f"• Treino: {len(train_df)} meses ({train_df['Data'].min().strftime('%Y-%m')} a {train_df['Data'].max().strftime('%Y-%m')})")
print(f"• Teste : {len(test_df)} meses ({test_df['Data'].min().strftime('%Y-%m')} a {test_df['Data'].max().strftime('%Y-%m')})")
display(pd.DataFrame(resultados_bench))
```

## 5. Passo 3: Benchmark *Zero-Shot* do Amazon Chronos (Famílias Bolt & T5 + Ensemble)

### O que vamos fazer agora e por quê?
Agora avaliamos **6 variantes pré-treinadas do Amazon Chronos** em modo **100% Zero-Shot** (sem nenhum treinamento ou ajuste de hiperparâmetros nos dados da EPE!) rodando em CPU:
- **Família Bolt (2ª geração, ultra-rápida)**: `chronos-bolt-tiny`, `chronos-bolt-mini`, `chronos-bolt-small`, `chronos-bolt-base`.
- **Família T5 (1ª geração, amostral)**: `chronos-t5-tiny`, `chronos-t5-small`.
- **Ensemble Chronos-Bolt (Mediana)**: Combina as previsões P10, P50 e P90 das 4 variantes Bolt para reduzir variância.

*Nota de Engenharia*: Nossa função auxiliar detecta automaticamente os pesos locais em `Models_cronos_amazon/cronos/` (execução instantânea offline) e faz fallback transparente para o **Hugging Face Hub** caso o repositório seja clonado em outra máquina — além de ser compatível tanto com `chronos-forecasting 1.x` quanto `2.x`!

```python
LOCAL_MODELS_DIR = Path("Models_cronos_amazon/cronos")

def prever_chronos_quantis(model_hf_id, serie_valores, horizonte, quantile_levels=(0.1, 0.5, 0.9)):
    '''Carrega modelo Chronos (local se disponível, senão Hugging Face) e prevê quantis.
    
    Compatível com chronos-forecasting 1.x (context=) e 2.x (inputs=).
    '''
    local_folder_name = model_hf_id.replace("/", "_")
    local_path = LOCAL_MODELS_DIR / local_folder_name

    if (local_path / "config.json").exists():
        source = str(local_path)
        local_only = True
    else:
        source = model_hf_id
        local_only = False

    t0 = time.perf_counter()
    pipe = BaseChronosPipeline.from_pretrained(
        source, device_map="cpu", dtype=torch.float32, local_files_only=local_only
    )

    ctx_tensor = torch.tensor(serie_valores, dtype=torch.float32).unsqueeze(0)  # [1, T]
    sig = inspect.signature(pipe.predict_quantiles)
    kwargs = {"prediction_length": horizonte, "quantile_levels": list(quantile_levels)}
    if "inputs" in sig.parameters:
        kwargs["inputs"] = ctx_tensor
    else:
        kwargs["context"] = ctx_tensor

    torch.manual_seed(42)
    qs, _ = pipe.predict_quantiles(**kwargs)
    q_np = qs[0].detach().cpu().numpy()  # [horizonte, 3]
    elapsed = time.perf_counter() - t0

    return q_np[:, 0], q_np[:, 1], q_np[:, 2], elapsed

modelos_chronos = [
    ("chronos-bolt-tiny",  "amazon/chronos-bolt-tiny",  "Chronos-Bolt"),
    ("chronos-bolt-mini",  "amazon/chronos-bolt-mini",  "Chronos-Bolt"),
    ("chronos-bolt-small", "amazon/chronos-bolt-small", "Chronos-Bolt"),
    ("chronos-bolt-base",  "amazon/chronos-bolt-base",  "Chronos-Bolt"),
    ("chronos-t5-tiny",    "amazon/chronos-t5-tiny",    "Chronos-T5"),
    ("chronos-t5-small",   "amazon/chronos-t5-small",   "Chronos-T5"),
]

for nome_curto, hf_id, familia in modelos_chronos:
    p10, p50, p90, dt = prever_chronos_quantis(hf_id, y_train, TEST_HORIZON)
    previsoes_teste[nome_curto] = {"p10": p10, "p50": p50, "p90": p90}
    resultados_bench.append(
        avaliar_previsao(nome_curto, familia, y_test, p50, p10, p90, tempo_s=dt)
    )
    print(f"✓ {nome_curto:<20} | MAPE: {resultados_bench[-1]['MAPE (%)']:5.2f}% | RMSE: {resultados_bench[-1]['RMSE (GWh)']:6.1f} GWh | Tempo: {dt:5.2f}s")

# Ensemble (Mediana das 4 variantes Chronos-Bolt)
bolt_keys = ["chronos-bolt-tiny", "chronos-bolt-mini", "chronos-bolt-small", "chronos-bolt-base"]
ens_p10 = np.median([previsoes_teste[k]["p10"] for k in bolt_keys], axis=0)
ens_p50 = np.median([previsoes_teste[k]["p50"] for k in bolt_keys], axis=0)
ens_p90 = np.median([previsoes_teste[k]["p90"] for k in bolt_keys], axis=0)
tempo_ens = sum(r["Tempo CPU (s)"] for r in resultados_bench if r["Modelo"] in bolt_keys)

previsoes_teste["Ensemble Chronos-Bolt"] = {"p10": ens_p10, "p50": ens_p50, "p90": ens_p90}
resultados_bench.append(
    avaliar_previsao("Ensemble Chronos-Bolt", "Ensemble Bolt", y_test, ens_p50, ens_p10, ens_p90, tempo_s=tempo_ens)
)

df_resultados = pd.DataFrame(resultados_bench).sort_values("MAPE (%)", ascending=True).reset_index(drop=True)
print("\n=== LEADERBOARD COMPLETO DE BACKTEST (AGO/2024 – JUL/2025, 12 MESES) ===")
display(df_resultados)
```

## 6. Passo 4: Visualização Comparativa — Backtest e Trade-off Acurácia vs. Velocidade

### O que vamos observar nos gráficos abaixo?
1. **Painel Esquerdo (Backtest Temporal)**: Compara visualmente os últimos 12 meses reais (Ago/2024 a Jul/2025) com o melhor modelo **Chronos**, o **Ensemble Bolt** e o baseline clássico **Holt-Winters**, destacando a banda probabilística **P10–P90**.
2. **Painel Direito (Acurácia vs. Custo Computacional)**: Mostra em um gráfico de dispersão qual família entrega o menor erro percentual (`MAPE %`) no menor tempo de inferência em CPU — destacando por que o **Chronos-Bolt** revoluciona o uso prático em produção sem GPU!

```python
melhor_chronos = df_resultados[df_resultados["Familia"].str.contains("Chronos")].iloc[0]["Modelo"]

fig, axes = plt.subplots(1, 2, figsize=(16, 5.5), gridspec_kw={"width_ratios": [1.45, 1.0]})

# 1. Backtest dos últimos 24 meses de histórico + 12 meses de teste
hist_recente = train_df.iloc[-24:]
axes[0].plot(hist_recente["Data"], hist_recente["consumo_BR_GWh"], color="#444444", lw=2.0, label="Histórico (Treino)")
axes[0].plot(test_df["Data"], y_test, color="black", lw=2.6, marker="o", markersize=5, label="Real (Teste 12m)")

# Melhor Chronos + faixa P10-P90
fc_best = previsoes_teste[melhor_chronos]
axes[0].plot(test_df["Data"], fc_best["p50"], color="#2ca02c", lw=2.4, linestyle="--", label=f"{melhor_chronos} (P50)")
axes[0].fill_between(test_df["Data"], fc_best["p10"], fc_best["p90"], color="#2ca02c", alpha=0.18, label=f"{melhor_chronos} (P10–P90)")

# Baseline Holt-Winters
axes[0].plot(test_df["Data"], previsoes_teste["Holt-Winters (ETS)"]["p50"], color="#d62728", lw=2.0, linestyle=":", label="Holt-Winters (ETS)")

axes[0].axvline(test_df["Data"].iloc[0], color="gray", linestyle="--", lw=1.1)
axes[0].set_title(f"Backtest (12 Meses): Real vs. {melhor_chronos} vs. Holt-Winters", fontsize=12.5, fontweight="bold")
axes[0].set_xlabel("Data")
axes[0].set_ylabel("Consumo Total Brasil (GWh)")
axes[0].legend(loc="upper left", fontsize=9, frameon=True)

# 2. Trade-off MAPE (%) vs. Tempo de Inferência em CPU (s)
cores_fam = {
    "Baseline Clássico": "#7f7f7f",
    "Chronos-Bolt": "#2ca02c",
    "Chronos-T5": "#d62728",
    "Ensemble Bolt": "#1f77b4",
}
for fam, grp in df_resultados.groupby("Familia"):
    axes[1].scatter(grp["Tempo CPU (s)"], grp["MAPE (%)"], s=110, color=cores_fam[fam], edgecolor="black", label=fam, zorder=5)
    for _, r in grp.iterrows():
        axes[1].annotate(
            r["Modelo"], (r["Tempo CPU (s)"], r["MAPE (%)"]),
            textcoords="offset points", xytext=(6, 4), fontsize=8.5
        )

axes[1].set_title("Trade-off em CPU: Erro Percentual (MAPE) vs. Latência", fontsize=12.5, fontweight="bold")
axes[1].set_xlabel("Tempo de Carregamento + Inferência em CPU (s) — Menor é Melhor")
axes[1].set_ylabel("MAPE (%) no Teste — Menor é Melhor")
axes[1].set_xlim(-0.5, df_resultados["Tempo CPU (s)"].max() * 1.25)
axes[1].legend(loc="upper right", fontsize=9, frameon=True)

plt.tight_layout()
plt.show()
```

## 7. Passo 5: Projeção Futura de 36 Meses (Ago/2025 – Jul/2028) com 100% do Histórico

### O que vamos fazer agora e por quê?
Depois de validar a acurácia no backtest fora da amostra, a etapa final de um projeto real de *Forecasting* é utilizar **todos os 259 meses disponíveis (Jan/2004 – Jul/2025)** como contexto para projetar os **próximos 36 meses (3 anos)** de demanda de energia elétrica no Brasil com o modelo campeão da família **Chronos-Bolt**, gerando projeções mensais **P10 (cenário conservador), P50 (cenário base) e P90 (cenário de alta demanda)** e salvando a tabela final em `outputs/projecao_energia_brasil_36m.csv`.

```python
HORIZONTE_FUTURO = 36
serie_completa = df["consumo_BR_GWh"].values.astype("float64")

# Usa o melhor modelo individual da família Chronos-Bolt (velocidade + acurácia)
melhor_bolt = df_resultados[df_resultados["Familia"] == "Chronos-Bolt"].iloc[0]["Modelo"]
hf_id_vencedor = f"amazon/{melhor_bolt}"

fut_p10, fut_p50, fut_p90, dt_fut = prever_chronos_quantis(hf_id_vencedor, serie_completa, HORIZONTE_FUTURO)
datas_futuras = pd.date_range(start=df["Data"].max() + pd.offsets.MonthBegin(1), periods=HORIZONTE_FUTURO, freq="MS")

df_projecao = pd.DataFrame({
    "Data": datas_futuras,
    "P10_Conservador_GWh": np.round(fut_p10, 1),
    "P50_Base_GWh": np.round(fut_p50, 1),
    "P90_Alto_GWh": np.round(fut_p90, 1),
})

# Salva projeção executiva em outputs/
Path("outputs").mkdir(exist_ok=True)
df_projecao.to_csv("outputs/projecao_energia_brasil_36m.csv", index=False, encoding="utf-8")

# Resumo anual projetado (média mensal em GWh)
resumo_anual = df_projecao.assign(Ano=df_projecao["Data"].dt.year).groupby("Ano")[
    ["P10_Conservador_GWh", "P50_Base_GWh", "P90_Alto_GWh"]
].mean().round(1)

print(f"• Modelo utilizado na projeção futura : {melhor_bolt} (Zero-Shot em {dt_fut:.2f} s)")
print("• Projeção salva em                   : outputs/projecao_energia_brasil_36m.csv")
print("\nMédia Mensal Projetada por Ano (GWh):")
display(resumo_anual)

# Gráfico executivo da projeção de 36 meses
fig, ax = plt.subplots(figsize=(13, 5.2))
hist_5a = df.iloc[-60:]
ax.plot(hist_5a["Data"], hist_5a["consumo_BR_GWh"], color="#1f77b4", lw=2.2, label="Histórico Real EPE (Últimos 5 Anos)")
ax.plot(df_projecao["Data"], df_projecao["P50_Base_GWh"], color="#2ca02c", lw=2.5, label=f"Projeção P50 — {melhor_bolt}")
ax.fill_between(
    df_projecao["Data"], df_projecao["P10_Conservador_GWh"], df_projecao["P90_Alto_GWh"],
    color="#2ca02c", alpha=0.20, label="Intervalo Probabilístico P10–P90"
)
ax.axvline(df["Data"].max(), color="gray", linestyle="--", lw=1.3)
ax.set_title(f"Projeção de 36 Meses do Consumo de Energia Elétrica no Brasil ({melhor_bolt} Zero-Shot)", fontsize=13, fontweight="bold")
ax.set_xlabel("Data")
ax.set_ylabel("Consumo Total (GWh)")
ax.legend(loc="upper left", frameon=True, fontsize=10)
plt.tight_layout()
plt.show()
```

## 8. Conclusões e Recomendações Práticas

1. **Força do *Zero-Shot* no Mundo Real**: Sem nenhum ajuste de parâmetros sazonais ou diferenciação manual, os *Foundation Models* **Amazon Chronos** capturaram com precisão tanto a tendência secular quanto os picos sazonais de verão do sistema elétrico brasileiro — superando ou rivalizando diretamente com o **Holt-Winters (ETS)** otimizado e esmagando o **Seasonal Naive**.
2. **`Chronos-Bolt` vs. `Chronos-T5` em Produção**: A família **Chronos-Bolt** entrega **acurácia de ponta com latência de poucos segundos em CPU**, eliminando a necessidade de GPUs caras para pipelines de previsão em lote.
3. **Valor da Previsão Probabilística (P10–P90)**: Para planejamento energético e dimensionamento de carga, a mediana (`P50`) sozinha não basta; a faixa **P10–P90** quantifica o risco de picos de demanda no verão (cenário `P90`) e desaceleração econômica (cenário `P10`).
4. **Aplicação Interativa Inclusa**: Este projeto também acompanha a interface web em Streamlit ([`app/App_Chronos.py`](app/App_Chronos.py)), atualizada para rodar de forma transparente em `chronos-forecasting 1.x` e `2.x`.
